We used the geographical coordinates.


```python
NYC_LAT_MIN = lat_min
NYC_LAT_MAX = lat_max

NYC_LON_MIN = lon_min
NYC_LON_MAX = lon_max
```

```python
NYC_POI_FILE = "/kaggle/working/nyc_pois.tsv"

first_chunk = True
total_nyc_pois = 0

with zipfile.ZipFile(ZIP_PATH, "r") as z:
    with z.open(POI_FILE) as f:

        for chunk in pd.read_csv(
            f,
            sep="\t",
            header=None,
            names=poi_columns,
            chunksize=500_000
        ):

            # Make sure coordinates are numeric
            chunk["latitude"] = pd.to_numeric(
                chunk["latitude"],
                errors="coerce"
            )

            chunk["longitude"] = pd.to_numeric(
                chunk["longitude"],
                errors="coerce"
            )

            mask = (
                chunk["latitude"].between(
                    NYC_LAT_MIN,
                    NYC_LAT_MAX
                )
                &
                chunk["longitude"].between(
                    NYC_LON_MIN,
                    NYC_LON_MAX
                )
            )

            nyc_chunk = chunk.loc[mask]

            if len(nyc_chunk) > 0:
                nyc_chunk.to_csv(
                    NYC_POI_FILE,
                    sep="\t",
                    index=False,
                    mode="w" if first_chunk else "a",
                    header=first_chunk
                )

                first_chunk = False
                total_nyc_pois += len(nyc_chunk)

print("NYC POIs:", total_nyc_pois)
```

```python
import os
import zipfile
import multiprocessing as mp

import pandas as pd
import numpy as np

# ============================================================
# CONFIG
# ============================================================

DATASET = "NYC"

ZIP_PATH = "/kaggle/working/lsbn2vec_global.zip"

OUTPUT_DIR = "/kaggle/working"

# Dataset #5 files
POI_FILE = "dataset_WWW2019/raw_POIs.txt"

CHECKIN_FILE = (
    "dataset_WWW2019/"
    "dataset_WWW_Checkins_anonymized.txt"
)

FRIENDSHIP_OLD_FILE = (
    "dataset_WWW2019/"
    "dataset_WWW_friendship_old.txt"
)

FRIENDSHIP_NEW_FILE = (
    "dataset_WWW2019/"
    "dataset_WWW_friendship_new.txt"
)

# ============================================================
# Same filtering as your previous TSMC notebook
# ============================================================

MIN_VENUE_VISITS = 10
MIN_USER_VISITS = 30

# ============================================================
# NYC geographic bounds
# ============================================================

NYC_LAT_MIN = 40.49
NYC_LAT_MAX = 40.92

NYC_LON_MIN = -74.27
NYC_LON_MAX = -73.65

# ============================================================
# Chunk sizes
# ============================================================

POI_CHUNKSIZE = 500_000
CHECKIN_CHUNKSIZE = 500_000
FRIENDSHIP_CHUNKSIZE = 500_000

os.makedirs(OUTPUT_DIR, exist_ok=True)

print("Configuration ready.")
```

## Load friendship edges and filter to our final active users

We only keep an edge if **both** users survived the POI/user visit filters above (i.e. both are in the final `df`).

```python
final_user_ids = set(df["user_id"].astype(str))
print(f"Final active users: {len(final_user_ids):,}")
```

```python
friendship_columns = ["user1", "user2"]

def load_and_filter_friendships(file_in_zip, valid_users):
    total_rows = 0
    total_kept = 0
    frames = []

    with zipfile.ZipFile(ZIP_PATH, "r") as z:
        with z.open(file_in_zip) as f:
            for chunk in pd.read_csv(
                f,
                sep="\t",
                header=None,
                names=friendship_columns,
                chunksize=FRIENDSHIP_CHUNKSIZE,
                dtype={"user1": str, "user2": str},
            ):
                total_rows += len(chunk)

                mask = (
                    chunk["user1"].isin(valid_users)
                    & chunk["user2"].isin(valid_users)
                )

                kept = chunk.loc[mask]

                if len(kept):
                    frames.append(kept)
                    total_kept += len(kept)

    print(file_in_zip)
    print(f"  Edges scanned : {total_rows:,}")
    print(f"  Edges kept (both users in our dataset): {total_kept:,}")

    if frames:
        return pd.concat(frames, ignore_index=True)
    return pd.DataFrame(columns=friendship_columns)

friends_old = load_and_filter_friendships(FRIENDSHIP_OLD_FILE, final_user_ids)
friends_new = load_and_filter_friendships(FRIENDSHIP_NEW_FILE, final_user_ids)
```

```python
friendships = pd.concat([friends_old, friends_new], ignore_index=True)

# de-duplicate edges (old + new files can overlap)
friendships = friendships.drop_duplicates().reset_index(drop=True)

# drop any self-loops just in case
friendships = friendships[friendships["user1"] != friendships["user2"]].copy()

n_users_in_friendships = pd.unique(friendships[["user1", "user2"]].values.ravel()).shape[0]

print("=" * 60)
print("FILTERED FRIENDSHIPS (only our final users)")
print("=" * 60)
print(f"Edges : {len(friendships):,}")
print(f"Users appearing in an edge: {n_users_in_friendships:,} / {len(final_user_ids):,}")

display(friendships.head())
```
